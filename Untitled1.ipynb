```python
import pandas as pd

class EcommerceAnalyzer:
    # Method 1: Properly indented inside the class
    def create_orders_df(self, data: list) -> pd.DataFrame:
        return pd.DataFrame(data, columns=["customerID", "Category", "OrderDate", "OrderStatus"])

    # Method 2: FIXED - Now correctly indented with a 4-space block inside the class
    def montly_delivery_rates(self, df: pd.DataFrame) -> pd.DataFrame:
        # step-1 -> copy original data
        temp = df.copy()

        # step-2 -> take only year-month
        temp["Month"] = temp["OrderDate"].str[:7]

        # step-3 -> count total orders (FIXED case-sensitivity: "customerID")
        total_orders = temp.groupby(["customerID", "Month"]).size().reset_index(name="TotalOrders")

        # step-4 -> count delivered orders (FIXED case-sensitivity: "customerID")
        delivered_orders = (temp[temp["OrderStatus"] == "Delivered"]
                            .groupby(["customerID", "Month"])
                            .size()
                            .reset_index(name="DeliveredOrders"))

        # step-5 -> merge aggregations (FIXED case-sensitivity: "customerID")
        result = total_orders.merge(delivered_orders, on=["customerID", "Month"], how="left")
        result["DeliveredOrders"] = result["DeliveredOrders"].fillna(0).astype(int)

        # step-6 -> calculate rate (FIXED variable typo: "DeliveredOrders" & column casing)
        result["Delivery Rate"] = (result["DeliveredOrders"] / result["TotalOrders"] * 100)

        return result[["customerID", "Month", "Delivery Rate"]]

    # method-3 -> Add return flag
    def add_return_flag(self,df:pd.DataFrame)->pd.DataFrame:
        result=df.copy()
        result["IsReturned"]=(result["OrderStatus"]=="Returned").astype(int)
        return result

    # method-4 -> frequent Returners
    def frequent_returners(self,df:pd.DataFrame,threshold: int)-> pd.DataFrame:
        result=(df[df["OrderStatus"]=="Returned"].groupby("customerID").size().reset_index(name="ReturnedCount"))

        return result[result["ReturnedCount"]>threshold]

    # method-5 -> category order summary
    def category_order_summary(self,df: pd.DataFrame) -> pd.DataFrame:

        summary = pd.crosstab(
            df["Category"],
            df["OrderStatus"]
        )

        summary = summary.reindex(
            columns=[
                "Delivered",
                "Cancelled",
                "Returned"
            ],
            fill_value=0
        )

        return summary.reset_index()


# ==========================================
# DRIVER CODE
# ==========================================

data = [
    [101, "Electronics", "2025-01-10", "Delivered"],
    [101, "Books", "2025-01-15", "returned"]
]

# 1. Instantiate the class to create an object
analyzer = EcommerceAnalyzer()

# 2. Call the method from the object to get the DataFrame
df = analyzer.create_orders_df(data)
print("--- Orders DataFrame ---")
print(df)
print("\n" + "="*40 + "\n")

# 3. FIXED: Passing 'df' instead of 'temp', calling the now-registered method
print("--- Monthly Delivery Rates ---")
print(analyzer.montly_delivery_rates(df))

print("---add Return flag---")
print(analyzer.add_return_flag(df))

print("--return count--")
print(analyzer.frequent_returners(df,1))

print("--category Orders--")
print(analyzer.category_order_summary(df))
```

```python
import pandas as pd

#hospital Visit Analyzer
class HospitalVisitAnalyzer:
    def create_visit_df(self,data:list) -> pd.DataFrame:
        return pd.DataFrame(data,columns=["PatientID","Department","VisitDate","VisitStatus"])

    def monthly_completion_rate(self,df:pd.DataFrame)-> pd.DataFrame:
        temp=df.copy()
        temp["Month"]=temp["VisitDate"].str[:7]

        total_visits = temp.groupby(["PatientID", "Month"]).size().reset_index(name="TotalVisits")

        completed_visits = (temp[temp["VisitStatus"].str.lower() == "completed"]
                            .groupby(["PatientID", "Month"])
                            .size()
                            .reset_index(name="CompletedVisits"))

        result = total_visits.merge(completed_visits, on=["PatientID", "Month"], how="left")
        result["CompletedVisits"] = result["CompletedVisits"].fillna(0).astype(int)
        result["Completion Rate"] = (result["CompletedVisits"] / result["TotalVisits"] * 100)

        return result[["PatientID", "Month", "Completion Rate"]]

    def add_cancellation_flag(self,df:pd.DataFrame)-> pd.DataFrame:

        result = df.copy()
        result["IsCancelled"] = (result["VisitStatus"].str.lower() == "cancelled").astype(int)
        return result

    def frequent_cancellers(self, df: pd.DataFrame, threshold: int) -> pd.DataFrame:
        result = (df[df["VisitStatus"].str.lower() == "cancelled"]
                  .groupby("PatientID")
                  .size()
                  .reset_index(name="CancellationCount"))
        return result[result["CancellationCount"] > threshold]

    # 5. Department Visit Summary
    def department_visit_summary(self, df: pd.DataFrame) -> pd.DataFrame:
        temp = df.copy()
        temp["VisitStatus"] = temp["VisitStatus"].str.capitalize()

        summary = pd.crosstab(temp["Department"], temp["VisitStatus"])
        summary = summary.reindex(columns=["Completed", "Cancelled", "Follow-up"], fill_value=0)
        return summary.reset_index()



analyzer=HospitalVisitAnalyzer()
data=[
    [501, "Cardiology", "2024-07-01", "Completed"],
    [502, "Neurology", "2024-07-01", "Cancelled"],
    [501, "Cardiology", "2024-07-10", "Follow-up"]
]

df=analyzer.create_visit_df(data)
print(df)

# monthly completion rate
print(analyzer.monthly_completion_rate(df))

print(analyzer.add_cancellation_flag(df))

print(analyzer.frequent_cancellers(df,1))

print(analyzer.department_visit_summary(df))
```

```python
# ==========================================
# Word Frequency Counter using OOP
# ==========================================
import re

class WordFrequencyCounter:
    """
    A class to analyze word frequencies in a sentence.
    """

    # ------------------------------------------
    # 1. Clean and Tokenize Input
    # ------------------------------------------
    def preprocess_text(self, text: str) -> list:
        """
        Converts text to lowercase,
        removes punctuation,
        and splits into words.
        """

        # Convert to lowercase
        text = text.lower()

        # Remove punctuation
        text = re.sub(r'[^\w\s]', '', text)

        # Tokenize words
        words = text.split()

        return words

    # ------------------------------------------
    # 2. Compute Word Frequency
    # ------------------------------------------
    def compute_word_frequency(self, words: list) -> dict:
        """
        Counts occurrences of each word.
        """

        frequency = {}

        for word in words:
            frequency[word] = frequency.get(word, 0) + 1

        return frequency

    # ------------------------------------------
    # 3. Get Most Frequent Word
    # ------------------------------------------
    def get_most_frequent_word(self, freq_dict: dict) -> tuple:
        """
        Returns the most frequent word and count.
        """

        if not freq_dict:
            return None

        return max(freq_dict.items(), key=lambda x: x[1])

    # ------------------------------------------
    # 4. Filter Words by Minimum Frequency
    # ------------------------------------------
    def filter_words_by_frequency(self, freq_dict: dict, n: int) -> dict:
        """
        Returns words appearing at least n times.
        """

        filtered_words = {}

        for word, count in freq_dict.items():
            if count >= n:
                filtered_words[word] = count

        return filtered_words


# ==========================================
# Main Program
# ==========================================

# Create object
counter = WordFrequencyCounter()

# Input sentence
sentence = """
Hello, hello! How are you?
I am fine. Hello Python!
Python is great and Python is powerful.
"""

# Step 1: Preprocess text
words = counter.preprocess_text(sentence)
print("Tokenized Words:")
print(words)

# Step 2: Compute frequency
frequency_dict = counter.compute_word_frequency(words)
print("\nWord Frequencies:")
print(frequency_dict)

# Step 3: Most frequent word
most_frequent = counter.get_most_frequent_word(frequency_dict)
print("\nMost Frequent Word:")
print(most_frequent)

# Step 4: Filter words with frequency >= 2
filtered_words = counter.filter_words_by_frequency(
    frequency_dict,
    2
)

print("\nWords Appearing At Least 2 Times:")
print(filtered_words)
```

```python
import pandas as pd

class EmployeeAttendanceAnalyzer:

    # 1. Create Attendance DataFrame
    def create_attendance_df(self, data: list) -> pd.DataFrame:
        """
        Converts raw attendance records into a structured DataFrame.
        """
        columns = ["EmployeeID", "Department", "Date", "AttendanceStatus"]
        df = pd.DataFrame(data, columns=columns)
        return df

    # 2. Calculate Monthly Presence Rate
    def monthly_presence_rate(self, df: pd.DataFrame) -> pd.DataFrame:
        """
        Calculates monthly presence rate for each employee.
        Presence Rate = (Present Days / Total Days) * 100
        """
        # Extract month
        df["Month"] = df["Date"].str[:7]

        # Total days per employee per month
        total_days = df.groupby(["EmployeeID", "Month"]).size().reset_index(name="TotalDays")

        # Present days per employee per month
        present_days = df[df["AttendanceStatus"] == "Present"].groupby(["EmployeeID", "Month"]).size().reset_index(name="PresentDays")

        # Merge total and present counts
        result = pd.merge(total_days, present_days, on=["EmployeeID", "Month"], how="left")
        result["PresentDays"] = result["PresentDays"].fillna(0)

        # Calculate presence rate
        result["Presence Rate"] = (result["PresentDays"] / result["TotalDays"]) * 100
        return result[["EmployeeID", "Month", "Presence Rate"]]

    # 3. Add Absence Flag
    def add_absence_flag(self, df: pd.DataFrame) -> pd.DataFrame:
        """
        Adds a flag column indicating if the attendance status was 'Absent'.
        """
        df["IsAbsent"] = (df["AttendanceStatus"] == "Absent").astype(int)
        return df

    # 4. Frequent Absentees
    def frequent_absentees(self, df: pd.DataFrame, threshold: int) -> pd.DataFrame:
        """
        Returns employees with absence count greater than threshold.
        """
        absent_df = df[df["AttendanceStatus"] == "Absent"]
        absence_count = absent_df.groupby("EmployeeID").size().reset_index(name="AbsenceCount")
        frequent = absence_count[absence_count["AbsenceCount"] > threshold]
        return frequent

    # 5. Department Attendance Summary
    def department_attendance_summary(self, df: pd.DataFrame) -> pd.DataFrame:
        """
        Returns department-wise summary of attendance statuses.
        """
        summary = pd.crosstab(df["Department"], df["AttendanceStatus"])

        # Ensure all expected columns exist
        for col in ["Present", "Absent", "Remote"]:
            if col not in summary.columns:
                summary[col] = 0

        summary = summary[["Present", "Absent", "Remote"]].reset_index()
        return summary


# ==============================
# Sample Data
# ==============================
data = [
    [101, "HR", "2024-06-01", "Present"],
    [102, "IT", "2024-06-01", "Absent"],
    [103, "Finance", "2024-06-01", "Present"],
    [101, "HR", "2024-06-02", "Remote"],
    [102, "IT", "2024-06-02", "Absent"],
    [103, "Finance", "2024-06-02", "Present"],
    [101, "HR", "2024-06-03", "Present"],
    [102, "IT", "2024-06-03", "Present"],
    [103, "Finance", "2024-06-03", "Remote"],
    [104, "Marketing", "2024-06-01", "Absent"],
    [104, "Marketing", "2024-06-02", "Absent"],
    [104, "Marketing", "2024-06-03", "Present"]
]

# ==============================
# Usage Example
# ==============================
analyzer = EmployeeAttendanceAnalyzer()

# 1. Create DataFrame
df = analyzer.create_attendance_df(data)
print("Attendance DataFrame:")
print(df, "\n")

# 2. Monthly Presence Rate
monthly_rate = analyzer.monthly_presence_rate(df)
print("Monthly Presence Rate:")
print(monthly_rate, "\n")

# 3. Add Absence Flag
df_with_flag = analyzer.add_absence_flag(df.copy())
print("DataFrame with Absence Flag:")
print(df_with_flag, "\n")

# 4. Frequent Absentees (threshold = 1)
frequent_absentees = analyzer.frequent_absentees(df, threshold=1)
print("Frequent Absentees (threshold = 1):")
print(frequent_absentees, "\n")

# 5. Department Attendance Summary
dept_summary = analyzer.department_attendance_summary(df)
print("Department Attendance Summary:")
print(dept_summary)
```
